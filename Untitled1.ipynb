```python
import numpy as np

text="birds can fly high"
words=text.split()
ids=[17,32,48,63]

emb={
17:[0.3,0.1,0.5,0.7],
32:[0.6,0.8,0.2,0.4],
48:[0.2,0.5,0.9,0.1],
63:[0.7,0.3,0.6,0.8]
}

x=np.array([emb[i] for i in ids])

pe=np.array([
[0.0,1.0,0.0,1.0],
[0.84,0.54,0.01,1.0],
[0.91,-0.42,0.02,1.0],
[0.14,-0.99,0.03,1.0]
])

z=x+pe

wq=np.array([
[1,0],
[0,1],
[0,1],
[1,0]
])

wk=np.array([
[0.5,1],
[1,0],
[0,0.5],
[1,0]
])

wv=np.array([
[1,0],
[0,0.5],
[0.5,1],
[0,1]
])

q=z@wq
k=z@wk
v=z@wv
s=q@k.T
s=s/np.sqrt(2)
e=np.exp(s-np.max(s,axis=1,keepdims=True))
a=e/np.sum(e,axis=1,keepdims=True)
o=a@v

wo=np.array([
[1,0,1,0],
[0,1,0,1]
])

h=o@wo
r=z+h
m=np.mean(r,axis=1,keepdims=True)
var=np.mean((r-m)**2,axis=1,keepdims=True)
y=(r-m)/np.sqrt(var+1e-8)

w1=np.array([
[1,0,1],
[0,1,1],
[1,-1,0],
[0,1,-1]
])

w2=np.array([
[1,0,1,0],
[0,1,0,1],
[1,1,0,1]
])

g1=y@w1
g=0.5*g1*(1+np.tanh(np.sqrt(2/np.pi)*(g1+0.044715*g1**3)))
f=g@w2
r2=y+f
m2=np.mean(r2,axis=1,keepdims=True)
var2=np.mean((r2-m2)**2,axis=1,keepdims=True)
out=(r2-m2)/np.sqrt(var2+1e-8)

np.set_printoptions(precision=4,suppress=True)

print("words:",words)
print("ids:",ids)
print("\nx:\n",x)
print("\npe:\n",pe)
print("\nz:\n",z)
print("\nq:\n",q)
print("\nk:\n",k)
print("\nv:\n",v)
print("\ns:\n",s)
print("\na:\n",a)
print("\no:\n",o)
print("\nh:\n",h)
print("\ny:\n",y)
print("\nf:\n",f)
print("\nout:\n",out)
```
